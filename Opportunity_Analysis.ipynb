```python
import pandas as pd
# Load the csv file into DataFrame
df = pd.read_csv("social_capital_college.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Data Cleaning
df.isnull().sum().sort_values(ascending=False)
```

```python
# Data Cleaning: Dropping columns with too many nulls and that are not relevant
cols_to_drop = [
    "ec_parent_ses_college", "ec_parent_ses_se_college", "ec_high_parent_ses_college",
    "ec_high_parent_ses_se_college", "bias_parent_ses_college", "bias_high_parent_ses_college",
    "exposure_parent_ses_college"
]
df = df.drop(columns=cols_to_drop)
```

```python
# Filling missing values using column Mean = smoothing
cols_with_missing = [
    "ec_own_ses_college", "exposure_own_ses_college", "bias_own_ses_college",
    "clustering_college", "support_ratio_college", "volunteering_rate_college", "mean_students_per_cohort"
]
df[cols_with_missing] = df[cols_with_missing].fillna(df[cols_with_missing].mean())
```

```python
# Data Cleaning: Rounding float values
df = df.round(2)
```

```python
df.head()
```

```python
df.tail()
```

```python
# We removed statistical error and subgroup-specific columns to simplify analysis, focusing on overall college-level SES dynamics.
cols_to_drop = [
    'ec_high_own_ses_se_college',
    'ec_high_own_ses_college',
    'ec_own_ses_se_college'
]
df.drop(columns=cols_to_drop, inplace=True)
```

```python
df.tail()
```

# Question 1 Which Colleges have the highest and lowest exposure to students from low-income backgrounds?

```python
df.sort_values(by='ec_own_ses_college', ascending=False)[['college_name', 'ec_own_ses_college']].head(10)
```

# Question 2 Is there a correlation between suppport_ratio_college and volunteering_rate_college?

```python
correlation = df['support_ratio_college'].corr(df['volunteering_rate_college'])
print(f"Correlation: {correlation}")
```

# Question 3 Which counties have the highest average social capital?

```python
# Step 1: Calculate average clustering_college (social capital) per county
avg_clustering_by_county = (
    df.groupby('county', as_index=False)['clustering_college']
    .mean()
    .rename(columns={'clustering_college': 'avg_clustering_college'})
)

# Step 2: Merge this average back into the original DataFrame
df_with_avg = df.merge(avg_clustering_by_county, on='county')

# Step 3: Sort by highest average clustering and then by college
top_county_colleges = df_with_avg[['county', 'avg_clustering_college', 'college_name', 'clustering_college']]
top_county_colleges = top_county_colleges.sort_values(by=['avg_clustering_college', 'clustering_college'], ascending=False)

# Step 4: Show top results (optional: top 20 or more depending on need)
top_county_colleges.head(10)

```

# Qusetion 4 What is the distrbution of bias_own_ses_college across colleges?

```python
import matplotlib.pyplot as plt
df['bias_own_ses_college'].hist(bins=30, edgecolor='black')
plt.title('Distribution of Own SES Bias')
plt.xlabel('Bias (own SES)')
plt.ylabel('Number of Colleges')
plt.show()
```

# Question 5 Do colleges with higher mean_students_per_cohort tend to have lower SES bias?

```python
df.plot.scatter(x='mean_students_per_cohort', y='bias_own_ses_college', alpha=0.6)
plt.title('SES Bias vs. Student Cohort Size')
plt.xlabel('Mean Students Per Cohort')
plt.ylabel('Own SES Bias')
plt.show()
```

# Question 6 How do volunteerong_rate_college and support_ratio_college vary by ZIP code?

```python
df.groupby('zip')[['volunteering_rate_college', 'support_ratio_college']].mean().sort_values(by='volunteering_rate_college', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Calculate average values by ZIP
zip_grouped = df.groupby('zip')[['volunteering_rate_college', 'support_ratio_college']].mean()

# Step 2: Select top ZIPs by volunteering rate
top_zips = zip_grouped.sort_values(by='volunteering_rate_college', ascending=False).tail(10)

# Step 3: Plot side-by-side bars
top_zips.plot(kind='bar', figsize=(12, 6))
plt.title('Average Volunteering Rate and Support Ratio by ZIP Code')
plt.xlabel('ZIP Code')
plt.ylabel('Average Rate')
plt.xticks(rotation=45)
plt.legend(['Volunteering Rate', 'Support Ratio'])
plt.tight_layout()
plt.show()
```
